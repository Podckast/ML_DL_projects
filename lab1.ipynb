```python
area_WPP = df[["Area (km²)", "World Population Percentage"]]
```

```python
sns.pairplot(area_WPP)
```

Распеределение WPP сильно смещено в левую сторону. Площадь страны слабо влияет на процент от популяции.


```python
sns.scatterplot(x=df['Growth Rate'], y = df["World Population Percentage"])
```

Корреляция слабая между Growth Rate и WPP

```python
sns.scatterplot(x=df['2022 Population'], y = df["World Population Percentage"])
```

В свою очередь корреляция между популяцией 2022 года и WPP положительная. И понятно почему: численное значение в столбце популяции 2022 года выражается в процентах относительно всей мировой популяции. Логично предположить, что другие столбцы Population тоже коррелируют с WPP

```python
sns.barplot(x = df['Continent'], y =df['World Population Percentage'])
```

Можно заметить, что Азия имеет самую большую долю в мировой популяции. На втором месте идет Южная Америка, далее Африка, Северная Америка и Европа. Океания имеет самую низкую долю.

```python
sns.histplot(df['Growth Rate'])
```

Коэффициент роста имеет распределение, близкое к нормальному. Однако имеется некоторая асимметрия, центральный столбец (Count = 60) резко выделяется. Также можно заметить показатели меньше 0.92, и больше 1.06, что удлиняет хвосты распределения - они могут быть потенциальными выбросами. 

```python
sns.boxplot(df['Area (km²)'])
```

График boxplot показывает, что большинство значений очень малы. Единицы значений - огромны, из-за этого медиана и ящик близки к нулю

```python
sns.boxplot(df['Density (per km²)'])
```

График boxplot плотности населения выглядит похожим образом. Медиана и ящик "прижаты" к нулю, а весь интерес сосредотачивается в выбросах.

```python
sns.violinplot(data=df, x ='World Population Percentage', y ='Continent')
```

Азия: подавляющее большинство азиатских стран имеют долю населения близкую к нулю, Китай и Индия создают длинный хвост.

Европа: большинство европейских стран имеют похожие доли населения. Плостность высокая в узком диапазоне. Нет длинных хвостов. 

Африка: похожа на Европу но немного более растянута вправо. 

Океаниия: все страны крошечные по населению, при этом внутри этого узкого диапазона стран много - высокая плотность, KDE раздувает ширину. 

Северная Америка: большинство стран Центральной Америки и Кариб - крошечные по населению, но США и Мексика создают хвост. 

Южная Америка: умеренная однородность, нет крайностей. Бразилия не создает гигантский выброс.
